```python
import os

# Main dataset folder
DATA_DIR = "/content/PlantVillage-Dataset/raw/color"

# Folder to access only tomato images
SELECTED_DIR = "/content/tomato_only"

# Create the folder
os.makedirs(SELECTED_DIR, exist_ok=True)

# Four tomato disease classes
tomato_classes = [
    "Tomato___healthy",
    "Tomato___Early_blight",
    "Tomato___Late_blight",
    "Tomato___Leaf_Mold"
]
```

```python
# Connect each original folder to our selected folder
for name in tomato_classes:
    source = os.path.join(DATA_DIR, name)
    destination = os.path.join(SELECTED_DIR, name)

    if not os.path.isdir(source):
        raise FileNotFoundError("Folder not found: " + name)

    if not os.path.exists(destination):
        os.symlink(source, destination)

print("Four tomato classes are ready!")
```

```python

# Image size
IMG_SIZE = 180

# Images processed at one time
BATCH_SIZE = 32

# Load training images
train_ds = keras.utils.image_dataset_from_directory(
    SELECTED_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)
```

```python
# Load validation images
val_ds = keras.utils.image_dataset_from_directory(
    SELECTED_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

# Store disease names
class_names = train_ds.class_names

print("Disease names:", class_names)
```

```python
from PIL import Image

for images, labels in train_ds.take(1):

    # Select one image
    image = images[0].numpy().astype("uint8")

    # Convert it into a Pillow image
    image = Image.fromarray(image)

    # Display the image
    plt.imshow(image)
    plt.title(class_names[labels[0].numpy()])
    plt.axis("off")
    plt.show()
```

```python
data_augmentation = keras.Sequential([
    keras.layers.RandomFlip("horizontal"),
    keras.layers.RandomRotation(0.1),
    keras.layers.RandomZoom(0.1)
])

print("Data augmentation ready!")
```

```python
model = keras.Sequential([

    # Input image
    keras.layers.Input(shape=(180, 180, 3)),

    # Change images slightly during training
    data_augmentation,

    # Scale pixel values between 0 and 1
    keras.layers.Rescaling(1./255),

    # CNN layers
    keras.layers.Conv2D(32, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D((2, 2)),


    keras.layers.Conv2D(64, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D((2, 2)),


    keras.layers.Conv2D(128, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D((2, 2)),

    # Convert features into one long array
    keras.layers.Flatten(),

    # Learn from the features
    keras.layers.Dense(128, activation="relu"),

    # Predict one of four classes
    keras.layers.Dense(4, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model is ready for training!")
```

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=7,
    callbacks=[early_stop]
)
```

```python
loss, accuracy = model.evaluate(val_ds)

print("Validation accuracy:", accuracy * 100, "%")
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

```python
def predict_disease(image):
    image = image.convert("RGB")
    image = image.resize((180, 180))

    image_array = np.array(image)
    image_array = np.expand_dims(image_array, axis=0)

    predictions = model.predict(image_array, verbose=0)[0]

    results = {
        class_names[i].replace("Tomato__", "").replace("", " "):
        float(predictions[i])
        for i in range(len(class_names))
    }

    return results

print("Prediction function ready!")
```

```python
app = gr.Interface(
    fn=predict_disease,
    inputs=gr.Image(type="pil", label="Upload Tomato Leaf"),
    outputs=gr.Label(num_top_classes=4),
    title="Tomato Disease Detection",
    description="Upload a tomato leaf image to predict its disease."
)

print("App interface ready!")
```
